## Loading Dataframes from files

```python
coffee = pd.read_csv('./warmup-data/coffee.csv')
bios = pd.read_csv('./data/bios.csv')
```

## Accessing Data

## Filtering Data

```python
bios.loc[(bios['height_cm'] > 215) & (bios['weight_kg'] > 100), ['name', 'height_cm', 'weight_kg']].sort_values(['height_cm', 'weight_kg'], ascending=[1,1]) #Filters the players with a height > 215 cm and a weight > 100 kg and sort both height and weight in ascending order
```

```python
bios.loc[(bios['name'].str.contains('Henry', case=False)) & (bios['born_country'] == 'USA')] # Filters for athletes with the word 'Henry' in them belonging to the USA
```

## Adding and Removing columns

```python
coffee['price'] = 5
```

```python
import numpy as np
coffee['new_price'] = np.where(coffee['Coffee Type'] == 'Espresso', 3, 6)
```

```python
coffee = coffee.drop(columns=['price'])
# Drops the column 'price' from the dataframe
```

```python
coffee = coffee.rename(columns={'new_price':'Updated Price'})
# Renames the column 'new_price' to 'Updated Price'
```
